```python
class_weights_array = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(y_full),
    y=y_full
)

class_weights = {i: weight for i, weight in enumerate(class_weights_array)}

print(f"Class 0 (Tidak Lanjut): {class_weights[0]:.4f}")
print(f"Class 1 (Lanjut):       {class_weights[1]:.4f}")
print(f"Weight Rasio: {class_weights[0]/class_weights[1]:.4f}:1")
```

```python
C_PARAM       = 0.1
PENALTY       = 'l2'
SOLVER        = 'liblinear'
MAX_ITER      = 2000
CV_FOLDS      = 10
CV_REPEATS    = 20
RANDOM_STATE  = 42

print(f"  Regularization (C):     {C_PARAM}")
print(f"  Penalty:                {PENALTY}")
print(f"  Solver:                 {SOLVER}")
print(f"  Max Iterations:         {MAX_ITER}")
print(f"  CV Folds:               {CV_FOLDS}")
print(f"  CV Repeats:             {CV_REPEATS}")
print(f"  Random State:           {RANDOM_STATE}")
```

```python
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('clf', LogisticRegression(
        C=C_PARAM,
        penalty=PENALTY,
        solver=SOLVER,
        max_iter=MAX_ITER,
        class_weight=class_weights,
        random_state=RANDOM_STATE
    ))
])

for step_name, step in pipeline.named_steps.items():
    print(f"  {step_name}: {step.__class__.__name__}")
```

```python
cv_oof = StratifiedKFold(n_splits=CV_FOLDS, shuffle=True, random_state=RANDOM_STATE)

oof_proba = cross_val_predict(
    pipeline, X_full, y_full,
    cv=cv_oof, method='predict_proba'
)[:, 1]

oof_pred = (oof_proba >= 0.5).astype(int)

print(f"OOF probabilitas shape: {oof_proba.shape}")
print(f"Probabilitas range:       [{oof_proba.min():.4f}, {oof_proba.max():.4f}]")
print(f"Mean probabilitas:        {oof_proba.mean():.4f}")
```

```python
cv_repeated = RepeatedStratifiedKFold(
    n_splits=CV_FOLDS, n_repeats=CV_REPEATS, random_state=RANDOM_STATE
)

cv_results = cross_validate(
    pipeline, X_full, y_full,
    cv=cv_repeated,
    scoring=['accuracy', 'precision', 'recall', 'f1', 'roc_auc', 'balanced_accuracy'],
    n_jobs=-1
)

print(f"Pengulangan CV ({CV_FOLDS}-fold × {CV_REPEATS} repeats = {CV_FOLDS*CV_REPEATS} evaluations)\n")
print(f"  Accuracy:          {cv_results['test_accuracy'].mean():.4f} ± {cv_results['test_accuracy'].std():.4f}")
print(f"  Precision:         {cv_results['test_precision'].mean():.4f} ± {cv_results['test_precision'].std():.4f}")
print(f"  Recall:            {cv_results['test_recall'].mean():.4f} ± {cv_results['test_recall'].std():.4f}")
print(f"  F1 Score:          {cv_results['test_f1'].mean():.4f} ± {cv_results['test_f1'].std():.4f}")
print(f"  ROC-AUC:           {cv_results['test_roc_auc'].mean():.4f} ± {cv_results['test_roc_auc'].std():.4f}")
print(f"  Balanced Accuracy: {cv_results['test_balanced_accuracy'].mean():.4f} ± {cv_results['test_balanced_accuracy'].std():.4f}")
```

```python
def evaluate_from_proba(y_true, y_proba, dataset_name='Dataset', threshold=0.5):

    y_pred = (y_proba >= threshold).astype(int)

    accuracy  = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall    = recall_score(y_true, y_pred, zero_division=0)
    f1        = f1_score(y_true, y_pred, zero_division=0)

    try:
        roc_auc = roc_auc_score(y_true, y_proba)
    except:
        roc_auc = 0.0

    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    tn, fp, fn, tp = cm.ravel()
    specificity  = tn / (tn + fp) if (tn + fp) > 0 else 0
    sensitivity  = tp / (tp + fn) if (tp + fn) > 0 else 0
    balanced_acc = (sensitivity + specificity) / 2.0

    metrics = {
        'accuracy':          accuracy,
        'precision':         precision,
        'recall':            recall,
        'f1_score':          f1,
        'roc_auc':           roc_auc,
        'balanced_accuracy': balanced_acc,
        'sensitivity':       sensitivity,
        'specificity':       specificity
    }

    print(f"{dataset_name} Performa (Threshold={threshold:.4f})")
    print(f"  Accuracy:          {accuracy:.4f}")
    print(f"  Precision:         {precision:.4f}")
    print(f"  Recall:            {recall:.4f}")
    print(f"  F1 Score:          {f1:.4f}")
    print(f"  ROC-AUC:           {roc_auc:.4f}")
    print(f"  Balanced Accuracy: {balanced_acc:.4f}\n")

    return metrics, cm
```

```python
insample_proba = final_model.predict_proba(X_full)[:, 1]

insample_metrics, insample_cm = evaluate_from_proba(
    y_full, insample_proba, 'IN-SAMPLE (Full Fit)'
)

oof_metrics, oof_cm = evaluate_from_proba(
    y_full, oof_proba, 'OUT-OF-FOLD (10-Fold CV)'
)
```

```python
fpr, tpr, thresholds = roc_curve(y_full, oof_proba)
optimal_idx = np.argmax(tpr - fpr)
optimal_threshold = thresholds[optimal_idx]

print(f"  Optimal Threshold:    {optimal_threshold:.4f}")
print(f"  TPR at optimal:       {tpr[optimal_idx]:.4f}")
print(f"  FPR at optimal:       {fpr[optimal_idx]:.4f}")
print(f"  Youdens J statistic: {tpr[optimal_idx] - fpr[optimal_idx]:.4f}\n")

oof_metrics_opt, oof_cm_opt = evaluate_from_proba(
    y_full, oof_proba, 'OUT-OF-FOLD (Optimal Threshold)', optimal_threshold
)
```
